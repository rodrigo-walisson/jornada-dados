## 1. Comparações e valores booleanos

Uma comparação sempre resulta em `True` (verdadeiro) ou `False` (falso).

| Operador | Significado | Excel |
|---|---|---|
| `==` | igual a | `=` |
| `!=` | diferente de | `<>` |
| `>` `<` | maior, menor | `>` `<` |
| `>=` `<=` | maior ou igual, menor ou igual | `>=` `<=` |

⚠️ **Atenção:** `=` **guarda** um valor numa variável. `==` **compara** dois valores. Confundir os dois é o erro mais comum desta semana.

**Exercício 1.** Preveja `True` ou `False` para cada linha **antes** de rodar. Duas delas costumam surpreender.

````python
print(10 > 5) #True
print(3 != 3) #False
print(10 == 10.0) #True
print("Brasil" == "brasil") #False
print("10" == 10) #False
print(7 >= 7) #True
````

## 2. Operadores lógicos: `and`, `or`, `not`

Servem para combinar condições, como `E()`, `OU()` e `NÃO()` no Excel.

| Operador | Resultado é `True` quando… | Excel |
|---|---|---|
| `and` | **as duas** condições são verdadeiras | `E()` |
| `or` | **pelo menos uma** é verdadeira | `OU()` |
| `not` | inverte: `True` vira `False` e vice-versa | `NÃO()` |

**Exercício 2.** Uma financeira aprova crédito apenas se a pessoa:
- tem renda de **pelo menos R$ 3.000**, **e**
- está com o **nome limpo**, **e**
- é **maior de idade** (18 anos ou mais).

Escreva a condição completa e guarde o resultado (`True` ou `False`) em `aprovado`.

````python
renda = 4500
nome_limpo = True
idade = 17

aprovado = renda >= 3000 and nome_limpo and idade >= 18
print(aprovado)
````

````python
confere(aprovado is False, "Dica: renda >= 3000 and nome_limpo and idade >= 18")
````

**Exercício 3.** Uma loja dá **frete grátis** se o cliente for **assinante** **ou** se a compra for **acima de R$ 200**.

Escreva a condição e guarde em `frete_gratis`.

````python
assinante = False
valor_compra = 250

frete_gratis = assinante or valor_compra > 200
print(frete_gratis)
````

````python
confere(frete_gratis is True, "Dica: assinante or valor_compra > 200")
````

## 3. `if` e `else`: tomando decisões

```python
if condicao:
    # roda se a condição for True
else:
    # roda se a condição for False
```

Três regras que o Python exige:
1. **Dois-pontos** `:` no final da linha do `if` e do `else`.
2. As linhas de dentro ficam **recuadas** (4 espaços, ou a tecla `Tab`). É o recuo que diz ao Python o que está "dentro" do `if`.
3. O `else` fica **alinhado** com o `if`.

**Exercício 4.** Calcule o **frete** e o **total** de uma compra:
- compra **acima de R$ 200** → frete R$ 0
- caso contrário → frete R$ 25

Use `if`/`else` (não vale escrever o valor do frete direto).

````python
valor_compra = 180

# escreva o if/else aqui, definindo a variável frete

if valor_compra > 200:
    frete = 0
else: 
    frete = 25
total = valor_compra + frete

print(f"Frete: R$ {frete:.2f} | Total: R$ {total:.2f}")
````

````python
confere(frete == 25, "Dica: if valor_compra > 200: frete = 0 / else: frete = 25")
confere(total == 205, "Dica: total = valor_compra + frete")
````

🔁 **Teste outros casos:** mude `valor_compra` para `250` e depois para exatamente `200`, e rode de novo. Os resultados fazem sentido? (Com `200`, o frete é cobrado ou não? Releia a regra: "**acima** de R$ 200".)

## 4. `elif`: mais de dois caminhos

Quando existem várias faixas, use `elif` (abreviação de *else if*). O Python testa de cima para baixo e **para na primeira condição verdadeira**.

```python
if condicao_1:
    ...
elif condicao_2:
    ...
else:
    ...
```

**Exercício 5.** A comissão de um vendedor depende do total vendido no mês:

| Vendas no mês | Comissão |
|---|---|
| até R$ 9.999,99 | 2% |
| de R$ 10.000 até R$ 29.999,99 | 4% |
| R$ 30.000 ou mais | 6% |

Calcule a comissão (em reais) para `vendas = 18500`.

````python
vendas = 18500

if vendas < 10000:
    comissao = vendas * 0.02
elif vendas < 30000:
    comissao = vendas * 0.04
else:
    comissao = vendas * 0.06

# escreva o if/elif/else aqui, definindo a variável comissao

print(f"Comissão: R$ {comissao:.2f}")
````

````python
confere(round(comissao, 2) == 740, "Dica: 18500 está na faixa de 4%  →  vendas * 0.04")
````

🔁 Teste com `vendas = 5000` (esperado: R$ 100,00) e `vendas = 30000` (esperado: R$ 1.800,00).

💡 **Pergunta para pensar:** se você escrever a condição `vendas >= 10000` **antes** de `vendas >= 30000`, o que acontece com quem vendeu R$ 50.000? Teste e veja.

**Exercício 6.** Pesquisas de satisfação usam o **NPS**: o cliente dá uma nota de 0 a 10 e é classificado assim:

| Nota | Classificação |
|---|---|
| 9 ou 10 | `"promotor"` |
| 7 ou 8 | `"neutro"` |
| 0 a 6 | `"detrator"` |
| fora de 0 a 10 | `"nota inválida"` |

Guarde a classificação (exatamente com esses textos) em `classificacao`.

````python
nota = 8

if nota in (9,10):
    classificacao = "promotor"
elif nota in (7,8):
    classificacao = "neutro"
elif 0 <= nota <= 6:
    classificacao = "detrator"
else:
    classificacao = "nota inválida"

print(classificacao)
````

````python
confere(classificacao == "neutro", "Dica: comece checando se a nota é inválida: if nota < 0 or nota > 10:")
````

🔁 Teste com `nota = 10`, `nota = 3` e `nota = 15`.

## 5. Do Excel para o Python

**Exercício 7.** Você encontrou esta fórmula numa planilha da escola:

```
=SE(B2>=7; "Aprovado"; SE(B2>=5; "Recuperação"; "Reprovado"))
```

Traduza para Python usando `if`/`elif`/`else`. A média está em `media` e o resultado vai em `situacao`.

Repare como o Python fica mais legível que vários `SE()` aninhados.

````python
media = 5.5

if media >= 7.0:
    situacao = "Aprovado"
elif media >= 5.0:
    situacao = "Recuperação"
else:
    situacao = "Reprovado"

print(situacao)
````

````python
confere(situacao == "Recuperação", "Dica: if media >= 7: ... elif media >= 5: ... else: ...")
````

## 6. Desafio: ano bissexto

Um ano é **bissexto** se:
- é divisível por **4**,
- **exceto** se também for divisível por **100**,
- **a não ser que** também seja divisível por **400**.

Exemplos: 2024 é bissexto. 1900 **não** é (divisível por 100). 2000 **é** (divisível por 400).

Escreva **uma única condição** (com `and`, `or` e `%`) e guarde em `bissexto`.

💡 Lembra do `%` da semana passada? "`ano` é divisível por 4" é o mesmo que `ano % 4 == 0`.

````python
ano = 2023

bissexto = ano % 4 == 0 and (ano % 100 != 0 or ano % 400 == 0)

print(bissexto)
````

````python
confere(bissexto is False, "Dica: ano % 4 == 0 and (ano % 100 != 0 or ano % 400 == 0)")
````

🔁 Teste com `2024` (True), `2023` (False) e `2000` (True).
