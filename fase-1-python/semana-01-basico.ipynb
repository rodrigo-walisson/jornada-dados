## 1. Python como calculadora

Assim como no Excel você escreve `=2+3*4`, no Python você escreve a expressão direto.

| Operador | O que faz | Exemplo | Resultado |
|---|---|---|---|
| `+ - * /` | soma, subtração, multiplicação, divisão | `7 / 2` | `3.5` |
| `//` | divisão inteira (descarta o resto) | `7 // 2` | `3` |
| `%` | resto da divisão | `7 % 2` | `1` |
| `**` | potência (no Excel é `^`) | `2 ** 3` | `8` |

A ordem de precedência é a mesma da matemática (e do Excel): parênteses → potência → multiplicação/divisão → soma/subtração.

**Exercício 1.** Preveja o resultado de cada linha **antes** de rodar. Depois rode e confira.

````python
print(2 + 3 * 4)
print((2 + 3) * 4)
print(17 // 5)
print(17 % 5)
print(2 ** 10)
````

**Exercício 2.** Uma empresa tem 47 funcionários e quer montar equipes de 6 pessoas.
- Quantas equipes **completas** dá para formar?
- Quantas pessoas **sobram**?

Use `//` e `%`.

````python
equipes_completas = 47//6
print(equipes_completas)
pessoas_sobrando = 47%6
print(pessoas_sobrando)
````

````python
confere(equipes_completas == 7, "Dica: 47 // 6")
confere(pessoas_sobrando == 5, "Dica: 47 % 6")
````

## 2. Tipos de dados

Todo valor no Python tem um **tipo**. Os quatro que você mais vai usar:

| Tipo | Nome | Exemplo | Parecido no Excel com… |
|---|---|---|---|
| `int` | número inteiro | `42` | número sem casas decimais |
| `float` | número decimal | `3.14` | número com casas decimais |
| `str` | texto (string) | `"Olá"` | célula de texto |
| `bool` | verdadeiro/falso | `True`, `False` | `VERDADEIRO`/`FALSO` |

⚠️ No Python o separador decimal é o **ponto**: `3.14`, nunca `3,14`.

Use `type()` para descobrir o tipo de um valor.

**Exercício 3.** Preveja o tipo de cada valor e depois rode.

````python
print(type(10))
print(type(10.0))
print(type("10"))
print(type(True))
````

## 3. Variáveis

Uma variável é um **nome que guarda um valor**, como dar nome a uma célula no Excel.

```python
preco = 65        # cria a variável preco com o valor 65
preco = preco + 5 # agora preco vale 70
```

**Boas práticas de nomes** (você vai ler muito código, então isso importa):
- letras minúsculas, palavras separadas por `_`: `preco_venda`, `total_mensal`
- nomes que digam o que é: `faturamento` é melhor que `f` ou `x`
- sem acentos e sem espaços

**Exercício 5.** Uma caneta custa R$ 40,00 para a empresa e é vendida por R$ 65,00. Calcule:
- o **lucro** em reais
- a **margem** em %, ou seja, `lucro / preço de venda × 100`

````python
preco_custo = 40
preco_venda = 65

lucro = preco_venda - preco_custo
margem =(lucro / preco_venda) * 100

print(lucro, margem)
````

````python
confere(lucro == 25, "Dica: preco_venda - preco_custo")
confere(margem is not None and round(margem, 2) == 38.46, "Dica: lucro / preco_venda * 100")
````

## 4. Textos (strings) e f-strings

Textos ficam entre aspas, simples `'assim'` ou duplas `"assim"`.

- `+` **junta** textos, como o `&` do Excel: `"Bom" + " dia"` → `"Bom dia"`
- `*` **repete**: `"-" * 10` → `"----------"`

A forma mais prática de misturar texto e variáveis é a **f-string**: coloque um `f` antes das aspas e as variáveis entre `{}`.

```python
nome = "Ana"
idade = 30
print(f"{nome} tem {idade} anos")   # Ana tem 30 anos
```

Para formatar números com casas decimais: `{valor:.2f}` (2 casas), `{valor:.1f}` (1 casa).

**Exercício 6.** Use uma f-string para montar exatamente este texto:

`Produto: Caneta | Margem: 38.5%`

Use as variáveis `produto` e `margem` (a margem precisa aparecer com **1 casa decimal**).

````python
produto = "Caneta"
margem = 25 / 65 * 100

frase =f"Produto: {produto} | Margem: {margem:.1f}%"
print(frase)
````

````python
confere(frase == "Produto: Caneta | Margem: 38.5%", 'Dica: f"Produto: {produto} | Margem: {margem:.1f}%"')
````

**Exercício 7.** Monte o nome completo juntando `nome` e `sobrenome` com um espaço no meio. No Excel seria `=A1&" "&B1`.

````python
nome = "Maria"
sobrenome = "Souza"

nome_completo = (nome) + " " + (sobrenome)
print(nome_completo)
````

````python
confere(nome_completo == "Maria Souza", 'Dica: nome + " " + sobrenome  (ou use uma f-string)')
````

## 5. Desafio: convertendo tempo

Um relatório de sistema mostra que um processo demorou **7384 segundos**. Ninguém entende isso numa reunião.

Converta para horas, minutos e segundos usando apenas `//` e `%`.

💡 Uma hora tem 3600 segundos e um minuto tem 60 segundos. Pense: quantas horas inteiras cabem? O que sobra? Quantos minutos inteiros cabem nessa sobra?

````python
total_segundos = 7384

horas = total_segundos // 3600
minutos = (total_segundos % 3600) // 60
segundos = total_segundos % 60

print(f"{horas}h {minutos}min {segundos}s")
````

````python
confere(horas == 2, "Dica: total_segundos // 3600")
confere(minutos == 3, "Dica: (total_segundos % 3600) // 60")
confere(segundos == 4, "Dica: total_segundos % 60")
````
