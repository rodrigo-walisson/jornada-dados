## 1. `for` com `range()`: repetir um número de vezes

```python
for i in range(5):
    print(i)
```

Leia assim: *"para cada `i` de 0 até 4, faça..."*. O que está **recuado** embaixo do `for` se repete.

| Forma | Gera | Observação |
|---|---|---|
| `range(5)` | 0, 1, 2, 3, 4 | começa em 0 e **para antes** do 5 |
| `range(1, 6)` | 1, 2, 3, 4, 5 | começa no 1º número, para antes do 2º |
| `range(0, 10, 2)` | 0, 2, 4, 6, 8 | o 3º número é o "passo" |

**Exercício 1.** Preveja o que cada laço vai imprimir **antes** de rodar. O mais traiçoeiro é o último.

````python
print("Laço A:")
for i in range(3):
    print(i)

print("Laço B:")
for mes in range(1, 4):
    print(f"Mês {mes}")

print("Laço C:")
for n in range(10, 0, -3):
    print(n)
````

## 2. O padrão acumulador

Este é **o padrão mais importante da semana**, e você já usou a ideia no simulador de desconto: começar com zero e ir somando.

### 📖 Exemplo resolvido
Somar os números de 1 a 5:

````python
total = 0                  # 1. começa com zero, ANTES do laço

for numero in range(1, 6):
    total = total + numero  # 2. soma um pouco a cada volta
    print(f"Somei {numero}, total parcial: {total}")

print(f"Total final: {total}")  # 3. usa o resultado DEPOIS do laço (sem recuo)
````

Repare nos três lugares:
1. a variável é criada **antes** do laço (fora dele);
2. ela é atualizada **dentro** do laço (com recuo);
3. o resultado é usado **depois** do laço (sem recuo).

Se o `total = 0` ficasse dentro do laço, ele seria zerado a cada volta. Esse é um erro muito comum.

**Exercício 2.** Some todos os números de **1 a 100** usando o padrão acumulador.

````python
confere(soma == 5050, "Dica: soma = 0 antes do laço, depois for n in range(1, 101): soma = soma + n")
````

## 3. Percorrendo uma coluna de dados

Até agora, o `for` percorreu números do `range()`. Mas ele também percorre **uma lista de valores**, como uma coluna do Excel. Você vai estudar listas a fundo na Semana 5; por enquanto, pense nelas como **uma coluna de valores entre colchetes**.

### 📖 Exemplo resolvido
Mostrar cada venda da semana e somar o total:

````python
vendas = [1200, 850, 3100, 470, 2200]

total = 0
for venda in vendas:          # "para cada venda na coluna vendas..."
    print(f"Venda: R$ {venda}")
    total = total + venda

print(f"Total da semana: R$ {total}")
````

````python
vendas = [1200, 850, 3100, 470, 2200]

acima_de_mil = None

# escreva o laço aqui

print(acima_de_mil)
````

## 4. `while`: repetir enquanto uma condição for verdadeira

O `for` é para quando você **sabe quantas vezes** vai repetir. O `while` é para quando você **não sabe**: ele repete **enquanto** a condição for verdadeira.

### 📖 Exemplo resolvido
Uma dívida de R$ 1.000 é paga em parcelas de R$ 300. Quantas parcelas são necessárias?

````python
divida = 1000
parcelas = 0

while divida > 0:           # "enquanto ainda houver dívida..."
    divida = divida - 300
    parcelas = parcelas + 1
    print(f"Parcela {parcelas}: falta R$ {max(divida, 0)}")

print(f"Total de parcelas: {parcelas}")
````

⚠️ **O perigo do `while`:** se nada dentro do laço mudar a condição, ele roda para sempre. Se você esquecesse a linha `divida = divida - 300`, a dívida nunca diminuiria. Laço infinito!

## 5. `break` e `continue`

| Comando | O que faz | Analogia |
|---|---|---|
| `break` | **sai** do laço imediatamente | achou o que procurava, para de procurar |
| `continue` | **pula** para a próxima volta | ignora esta linha e segue para a próxima |

### 📖 Exemplo resolvido

````python
print("Com break: para no primeiro número maior que 3")
for n in range(1, 10):
    if n > 3:
        break
    print(n)

print("Com continue: pula os números pares")
for n in range(1, 8):
    if n % 2 == 0:
        continue
    print(n)
````
